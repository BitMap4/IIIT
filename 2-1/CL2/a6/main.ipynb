# With k=200

```python
config = AnalysisConfig(
    window_sizes=[2, 5, 8],
    num_components=200
)

analyzer = WordEmbeddingAnalyzer('train.csv', config)

# custom_pairs = [
#     WordPair("market", "stock", "related"),
#     WordPair("price", "cost", "synonym"),
#     WordPair()
# ]
# results = analyzer.analyze(custom_pairs)

results = analyzer.analyze()
analyzer.print_results()
```

# With k=50

```python
config = AnalysisConfig(window_sizes=[2, 5, 8])

analyzer = WordEmbeddingAnalyzer('train.csv', config)
analyzer.analyze()
analyzer.print_results()
```

# Writing matrix to matrix.csv with context window of size 5 and k=50

```python
config = AnalysisConfig(
    max_lines=int(0.15*120000),
    window_sizes=[2, 5]
)

analyzer = WordEmbeddingAnalyzer('train.csv', config)

analyzer.analyze()
analyzer.print_results()
analyzer.cooccurrence.print_matrix("matrix.csv")
```
